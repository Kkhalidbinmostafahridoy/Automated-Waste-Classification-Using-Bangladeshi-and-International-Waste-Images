```python
def show_tree(path, max_depth=3, current_depth=0):

    path = Path(path)

    if current_depth > max_depth:
        return

    if not path.exists():
        print("❌ Not found:", path)
        return

    indent = "    " * current_depth

    for item in sorted(
        path.iterdir(),
        key=lambda x: (not x.is_dir(), x.name.lower())
    ):

        print(
            indent +
            ("📁 " if item.is_dir() else "📄 ") +
            item.name
        )

        if item.is_dir():

            show_tree(
                item,
                max_depth,
                current_depth + 1
            )


show_tree(
    INTERNATIONAL_DATA,
    max_depth=3
)
```

```python
from pathlib import Path
import shutil
import random
import hashlib
from collections import defaultdict

INTERNATIONAL_DATA = Path(
    "/content/drive/MyDrive/Waste_Dataset/International_data"
)

TARGET_CLASSES = [
    "Cardboard",
    "Glass",
    "Metal",
    "Paper",
    "Plastic"
]

IMAGE_EXTENSIONS = {
    ".jpg",
    ".jpeg",
    ".png",
    ".webp",
    ".bmp"
}

print("International dataset:")
print(INTERNATIONAL_DATA)

for class_name in TARGET_CLASSES:

    folder = INTERNATIONAL_DATA / class_name

    print(
        f"{class_name:12s}: "
        f"{folder.exists()}"
    )
```

```python
all_international_images = {}

print("=" * 80)
print("COLLECTING INTERNATIONAL IMAGES")
print("=" * 80)

for class_name in TARGET_CLASSES:

    class_folder = (
        INTERNATIONAL_DATA /
        class_name
    )

    images = []

    for image_path in class_folder.rglob("*"):

        if (
            image_path.is_file()
            and
            image_path.suffix.lower()
            in IMAGE_EXTENSIONS
        ):

            images.append(image_path)

    all_international_images[
        class_name
    ] = images

    print(
        f"{class_name:12s}: "
        f"{len(images):5d}"
    )
```

```python
def calculate_hash(path):

    md5 = hashlib.md5()

    with open(path, "rb") as f:

        for chunk in iter(
            lambda: f.read(1024 * 1024),
            b""
        ):
            md5.update(chunk)

    return md5.hexdigest()
```

```python
# ============================================================
# COLLECT UNIQUE INTERNATIONAL IMAGES
# ============================================================

clean_international_images = {}

print("=" * 80)
print("COLLECTING CLEAN INTERNATIONAL IMAGES")
print("=" * 80)

grand_before = 0
grand_after = 0
grand_duplicates = 0


for class_name in TARGET_CLASSES:

    source_dir = (
        INTERNATIONAL_DATA /
        class_name
    )

    images = [
        p
        for p in source_dir.rglob("*")
        if (
            p.is_file()
            and
            p.suffix.lower()
            in IMAGE_EXTENSIONS
        )
    ]

    seen_hashes = set()
    unique_images = []

    duplicate_count = 0

    for image_path in images:

        image_hash = calculate_hash(
            image_path
        )

        if image_hash in seen_hashes:

            duplicate_count += 1
            continue

        seen_hashes.add(image_hash)

        unique_images.append(
            image_path
        )

    clean_international_images[
        class_name
    ] = unique_images

    grand_before += len(images)
    grand_after += len(unique_images)
    grand_duplicates += duplicate_count

    print(
        f"{class_name:12s} | "
        f"Before: {len(images):5d} | "
        f"Duplicates: {duplicate_count:3d} | "
        f"After: {len(unique_images):5d}"
    )


print("-" * 80)

print(
    "TOTAL BEFORE:",
    grand_before
)

print(
    "TOTAL DUPLICATES:",
    grand_duplicates
)

print(
    "TOTAL AFTER:",
    grand_after
)
```

```python
def create_split(
    images,
    train_ratio=0.80,
    valid_ratio=0.10,
    seed=42
):

    images = list(images)

    random.Random(seed).shuffle(
        images
    )

    total = len(images)

    train_count = int(
        total * train_ratio
    )

    valid_count = int(
        total * valid_ratio
    )

    train_images = images[
        :train_count
    ]

    valid_images = images[
        train_count:
        train_count + valid_count
    ]

    test_images = images[
        train_count + valid_count:
    ]

    return {
        "train": train_images,
        "valid": valid_images,
        "test": test_images
    }
```

```python
# ============================================================
# CREATE PROCESSED INTERNATIONAL DATASET
# ============================================================

# Define the path for the processed international dataset
PROCESSED_INT = Path("/content/drive/MyDrive/Waste_Dataset/Processed_International_Data")

# Create main directory
PROCESSED_INT.mkdir(
    parents=True,
    exist_ok=True
)

# Create train/valid/test/class directories
for split in ["train", "valid", "test"]:

    for class_name in TARGET_CLASSES:

        (
            PROCESSED_INT /
            split /
            class_name
        ).mkdir(
            parents=True,
            exist_ok=True
        )


FINAL_INT_STATS = {}

print("=" * 80)
print("CREATING PROCESSED INTERNATIONAL DATASET")
print("=" * 80)


for class_name in TARGET_CLASSES:

    images = clean_international_images[
        class_name
    ]

    splits = create_split(
        images,
        train_ratio=0.80,
        valid_ratio=0.10,
        seed=42
    )

    FINAL_INT_STATS[
        class_name
    ] = {}

    print(
        f"\n{class_name}"
    )

    for split, split_images in splits.items():

        destination_dir = (
            PROCESSED_INT /
            split /
            class_name
        )

        for index, source_path in enumerate(
            split_images,
            start=1
        ):

            destination = (
                destination_dir /
                f"INT_{class_name}_"
                f"{index:06d}"
                f"{source_path.suffix.lower()}"
            )

            shutil.copy2(
                source_path,
                destination
            )

        FINAL_INT_STATS[
            class_name
        ][split] = len(
            split_images
        )

        print(
            f"  {split:5s}: "
            f"{len(split_images)}"
        )


print("\n✅ INTERNATIONAL PROCESSED DATASET CREATED")
```

```python
def count_images(path):

    path = Path(path)

    if not path.exists():
        return 0

    return sum(
        1
        for p in path.rglob("*")
        if (
            p.is_file()
            and
            p.suffix.lower()
            in IMAGE_EXTENSIONS
        )
    )
```

```python
print("=" * 80)
print("PROCESSED INTERNATIONAL DATASET")
print("=" * 80)

grand_total = 0

for split in [
    "train",
    "valid",
    "test"
]:

    print("\n" + "=" * 60)
    print(split.upper())
    print("=" * 60)

    split_total = 0

    for class_name in TARGET_CLASSES:

        folder = (
            PROCESSED_INT /
            split /
            class_name
        )

        count = count_images(
            folder
        )

        print(
            f"{class_name:12s}: "
            f"{count:5d}"
        )

        split_total += count

    print("-" * 60)

    print(
        f"TOTAL       : "
        f"{split_total:5d}"
    )

    grand_total += split_total


print("\n" + "=" * 60)

print(
    f"GRAND TOTAL : {grand_total}"
)

print("=" * 60)
```

```python
# ============================================================
# FINAL INTERNATIONAL DUPLICATE CHECK
# ============================================================

hash_groups = defaultdict(list)


for split in [
    "train",
    "valid",
    "test"
]:

    for class_name in TARGET_CLASSES:

        folder = (
            PROCESSED_INT /
            split /
            class_name
        )

        for image_path in folder.rglob("*"):

            if (
                image_path.is_file()
                and
                image_path.suffix.lower()
                in IMAGE_EXTENSIONS
            ):

                image_hash = calculate_hash(
                    image_path
                )

                hash_groups[
                    image_hash
                ].append(
                    image_path
                )


duplicates = {
    h: paths
    for h, paths in hash_groups.items()
    if len(paths) > 1
}


print("=" * 80)
print("FINAL INTERNATIONAL DUPLICATE CHECK")
print("=" * 80)

print(
    "Total images:",
    sum(
        len(v)
        for v in hash_groups.values()
    )
)

print(
    "Unique hashes:",
    len(hash_groups)
)

print(
    "Duplicate groups:",
    len(duplicates)
)
```

```python
# ============================================================
# CROSS-SPLIT LEAKAGE CHECK
# ============================================================

leakage_groups = []


for image_hash, paths in hash_groups.items():

    splits_found = set()

    for path in paths:

        parts = path.parts

        if "train" in parts:
            splits_found.add("train")

        elif "valid" in parts:
            splits_found.add("valid")

        elif "test" in parts:
            splits_found.add("test")


    if len(splits_found) > 1:

        leakage_groups.append(
            (
                image_hash,
                paths,
                splits_found
            )
        )


print("=" * 80)
print("INTERNATIONAL CROSS-SPLIT LEAKAGE CHECK")
print("=" * 80)

print(
    "Cross-split duplicate groups:",
    len(leakage_groups)
)
```
