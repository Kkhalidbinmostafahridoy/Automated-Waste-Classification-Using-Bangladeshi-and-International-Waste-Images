```python
def get_images(folder):

    folder = Path(folder)

    if not folder.exists():
        return []

    return [
        p
        for p in folder.rglob("*")
        if (
            p.is_file()
            and
            p.suffix.lower()
            in IMAGE_EXTENSIONS
        )
    ]


def count_images(folder):

    return len(
        get_images(folder)
    )
```

```python
def normalize_class_name(name):

    original = str(name).strip()

    name = original.lower()

    name = name.replace("_", " ")
    name = name.replace("-", " ")

    name = " ".join(
        name.split()
    )


    # Cardboard
    if "cardboard" in name:
        return "Cardboard"


    # Glass
    if "glass" in name:
        return "Glass"


    # Metal
    if "metal" in name:
        return "Metal"


    # Paper
    if "paper" in name:
        return "Paper"

    if "newspaper" in name:
        return "Paper"


    # Plastic
    if "plastic" in name:
        return "Plastic"

    if "polythene" in name:
        return "Plastic"


    # Extra classes
    return None
```

```python
def process_root_dataset(
    source_dir,
    source_name,
    output_dir,
    train_ratio=0.80,
    valid_ratio=0.10,
    test_ratio=0.10,
    seed=42
):

    source_dir = Path(source_dir)
    output_dir = Path(output_dir)

    print("\n")
    print("=" * 75)
    print(
        f"PROCESSING: {source_name}"
    )
    print("=" * 75)


    rng = random.Random(seed)


    statistics = {}


    for target_class in TARGET_CLASSES:

        all_images = []


        # ----------------------------------------------------
        # Read every folder inside dataset
        # ----------------------------------------------------

        for folder in source_dir.iterdir():

            if not folder.is_dir():
                continue


            mapped_class = normalize_class_name(
                folder.name
            )


            # Extra class
            if mapped_class is None:
                continue


            if mapped_class != target_class:
                continue


            images = get_images(
                folder
            )


            all_images.extend(
                images
            )


        # Remove duplicate paths
        all_images = list(
            dict.fromkeys(
                all_images
            )
        )


        # Shuffle
        rng.shuffle(
            all_images
        )


        total = len(
            all_images
        )


        train_count = int(
            total * train_ratio
        )

        valid_count = int(
            total * valid_ratio
        )

        test_count = (
            total
            - train_count
            - valid_count
        )


        train_images = all_images[
            :train_count
        ]

        valid_images = all_images[
            train_count:
            train_count + valid_count
        ]

        test_images = all_images[
            train_count + valid_count:
        ]


        splits = {
            "train": train_images,
            "valid": valid_images,
            "test": test_images
        }


        statistics[target_class] = {}


        for split, images in splits.items():

            destination_dir = (
                output_dir /
                split /
                target_class
            )

            destination_dir.mkdir(
                parents=True,
                exist_ok=True
            )


            for index, image_path in enumerate(
                images,
                start=1
            ):

                extension = (
                    image_path.suffix.lower()
                )


                filename = (
                    f"{source_name}_"
                    f"{target_class}_"
                    f"{index:06d}"
                    f"{extension}"
                )


                destination = (
                    destination_dir /
                    filename
                )


                shutil.copy2(
                    image_path,
                    destination
                )


            statistics[
                target_class
            ][split] = len(images)


        print(
            f"{target_class:12s} | "
            f"Total={total:5d} | "
            f"Train={train_count:5d} | "
            f"Valid={valid_count:5d} | "
            f"Test={test_count:5d}"
        )


    return statistics
```

```python
def read_yolo_labels(label_file):

    objects = []

    if not label_file.exists():
        return objects

    with open(
        label_file,
        "r",
        encoding="utf-8"
    ) as f:

        for line in f:

            line = line.strip()

            if not line:
                continue

            parts = line.split()

            try:

                class_id = int(parts[0])

                objects.append(class_id)

            except:

                continue

    return objects
```

```python
EWASTE_CLASS_NAMES = [
    "Battery_Waste",
    "Glass_Waste",
    "Keyboard",
    "Light_Bulb",
    "Medical_Waste",
    "Metal_Waste",
    "Mobile",
    "Mouse",
    "Organic_Waste",
    "PCB",
    "Paper_Waste",
    "Plastic_Waste"
]


from collections import Counter


for split in ["train", "valid", "test"]:

    label_dir = (
        EWASTE_RAW /
        split /
        "labels"
    )

    counter = Counter()

    if label_dir.exists():

        for label_file in label_dir.glob("*.txt"):

            ids = read_yolo_labels(
                label_file
            )

            counter.update(ids)


    print("\n" + "=" * 70)
    print(split.upper())
    print("=" * 70)


    for class_id, class_name in enumerate(
        EWASTE_CLASS_NAMES
    ):

        print(
            f"{class_id:2d} | "
            f"{class_name:20s} | "
            f"{counter[class_id]}"
        )
```

```python
EWASTE_CLASS_MAPPING = {
    1: "Glass",
    5: "Metal",
    10: "Paper",
    11: "Plastic"
}
```

```python
def process_ewaste_dataset(
    source_dir,
    output_dir
):

    source_dir = Path(source_dir)
    output_dir = Path(output_dir)

    stats = {
        "train": Counter(),
        "valid": Counter(),
        "test": Counter()
    }


    for split in [
        "train",
        "valid",
        "test"
    ]:

        images_dir = (
            source_dir /
            split /
            "images"
        )

        labels_dir = (
            source_dir /
            split /
            "labels"
        )


        if not images_dir.exists():

            print(
                f"⚠️ Images not found: {images_dir}"
            )

            continue


        print("\n")
        print("=" * 70)
        print(
            f"PROCESSING E-WASTE: {split.upper()}"
        )
        print("=" * 70)


        for image_path in images_dir.iterdir():

            if not image_path.is_file():
                continue


            if (
                image_path.suffix.lower()
                not in IMAGE_EXTENSIONS
            ):
                continue


            label_path = (
                labels_dir /
                f"{image_path.stem}.txt"
            )


            if not label_path.exists():

                continue


            class_ids = read_yolo_labels(
                label_path
            )


            # ------------------------------------------------
            # Find target classes inside this image
            # ------------------------------------------------

            target_classes = []


            for class_id in class_ids:

                if class_id in EWASTE_CLASS_MAPPING:

                    target_class = (
                        EWASTE_CLASS_MAPPING[
                            class_id
                        ]
                    )

                    target_classes.append(
                        target_class
                    )


            # Remove duplicates
            target_classes = list(
                dict.fromkeys(
                    target_classes
                )
            )


            # ------------------------------------------------
            # Copy image to each relevant class
            # ------------------------------------------------

            for target_class in target_classes:

                destination_dir = (
                    output_dir /
                    split /
                    target_class
                )


                destination_dir.mkdir(
                    parents=True,
                    exist_ok=True
                )


                destination = (
                    destination_dir /
                    f"E_WASTE_{image_path.name}"
                )


                # Prevent filename collision
                if destination.exists():

                    destination = (
                        destination_dir /
                        f"E_WASTE_{image_path.stem}_"
                        f"{abs(hash(str(image_path)))}"
                        f"{image_path.suffix.lower()}"
                    )


                shutil.copy2(
                    image_path,
                    destination
                )


                stats[
                    split
                ][target_class] += 1


    return stats
```

```python
import hashlib
from collections import defaultdict

IMAGE_EXTENSIONS = {
    ".jpg",
    ".jpeg",
    ".png",
    ".webp",
    ".bmp"
}


def file_hash(path):

    hash_md5 = hashlib.md5()

    with open(path, "rb") as f:

        for chunk in iter(
            lambda: f.read(1024 * 1024),
            b""
        ):
            hash_md5.update(chunk)

    return hash_md5.hexdigest()


hash_groups = defaultdict(list)

for split in ["train", "valid", "test"]:

    for class_name in TARGET_CLASSES:

        folder = (
            UNIFIED_BD /
            split /
            class_name
        )

        if not folder.exists():
            continue

        for image_path in folder.rglob("*"):

            if (
                image_path.is_file()
                and
                image_path.suffix.lower()
                in IMAGE_EXTENSIONS
            ):

                h = file_hash(image_path)

                hash_groups[h].append(
                    image_path
                )


duplicates = {
    h: paths
    for h, paths in hash_groups.items()
    if len(paths) > 1
}


print("=" * 80)
print("DUPLICATE IMAGE CHECK")
print("=" * 80)

print(
    "Unique image hashes:",
    len(hash_groups)
)

print(
    "Duplicate groups:",
    len(duplicates)
)
```

```python
print("=" * 80)
print("DUPLICATE DETAILS")
print("=" * 80)

duplicate_image_count = 0

for i, (h, paths) in enumerate(
    duplicates.items(),
    start=1
):

    print("\n" + "-" * 80)
    print(f"Duplicate Group #{i}")
    print("-" * 80)

    for path in paths:

        print(path)

    duplicate_image_count += len(paths)


print("\n" + "=" * 80)
print(
    "Duplicate groups:",
    len(duplicates)
)

print(
    "Images involved in duplicate groups:",
    duplicate_image_count
)
```

```python
print("=" * 80)
print("DUPLICATE SPLIT LEAKAGE CHECK")
print("=" * 80)

leakage_groups = []

for h, paths in duplicates.items():

    splits_found = set()

    for path in paths:

        parts = path.parts

        if "train" in parts:
            splits_found.add("train")

        elif "valid" in parts:
            splits_found.add("valid")

        elif "test" in parts:
            splits_found.add("test")


    if len(splits_found) > 1:

        leakage_groups.append(
            (h, paths, splits_found)
        )


print(
    "Cross-split duplicate groups:",
    len(leakage_groups)
)
```

```python
import hashlib
import random
import shutil
from pathlib import Path
from collections import defaultdict


IMAGE_EXTENSIONS = {
    ".jpg",
    ".jpeg",
    ".png",
    ".webp",
    ".bmp"
}


def calculate_hash(path):

    md5 = hashlib.md5()

    with open(path, "rb") as f:

        for chunk in iter(
            lambda: f.read(1024 * 1024),
            b""
        ):

            md5.update(chunk)

    return md5.hexdigest()
```

```python
def remove_duplicate_paths(image_paths):

    unique_images = []

    seen_hashes = set()

    duplicate_count = 0


    for image_path in image_paths:

        image_hash = calculate_hash(
            image_path
        )


        if image_hash in seen_hashes:

            duplicate_count += 1

            continue


        seen_hashes.add(
            image_hash
        )

        unique_images.append(
            image_path
        )


    return unique_images, duplicate_count
```

```python
def create_clean_split(
    images,
    seed=42
):

    images = list(images)

    random.Random(seed).shuffle(
        images
    )


    total = len(images)


    train_count = int(
        total * 0.80
    )

    valid_count = int(
        total * 0.10
    )


    train_images = images[
        :train_count
    ]


    valid_images = images[
        train_count:
        train_count + valid_count
    ]


    test_images = images[
        train_count + valid_count:
    ]


    return {
        "train": train_images,
        "valid": valid_images,
        "test": test_images
    }
```

```python
FINAL_STATS = {}


for class_name in TARGET_CLASSES:

    images = clean_class_images[
        class_name
    ]


    splits = create_clean_split(
        images,
        seed=42
    )


    FINAL_STATS[
        class_name
    ] = {}


    for split, split_images in splits.items():

        destination_dir = (
            CLEAN_UNIFIED_BD /
            split /
            class_name
        )


        for index, source_path in enumerate(
            split_images,
            start=1
        ):

            destination = (
                destination_dir /
                f"BD_{class_name}_"
                f"{index:06d}"
                f"{source_path.suffix.lower()}"
            )


            shutil.copy2(
                source_path,
                destination
            )


        FINAL_STATS[
            class_name
        ][split] = len(
            split_images
        )


print("✅ Clean Bangladesh dataset created")
```

```python
hash_groups = defaultdict(list)


for split in [
    "train",
    "valid",
    "test"
]:

    for class_name in TARGET_CLASSES:

        folder = (
            CLEAN_UNIFIED_BD /
            split /
            class_name
        )


        for image_path in folder.rglob("*"):

            if (
                image_path.is_file()
                and
                image_path.suffix.lower()
                in IMAGE_EXTENSIONS
            ):

                h = calculate_hash(
                    image_path
                )

                hash_groups[h].append(
                    image_path
                )


duplicates = {
    h: paths
    for h, paths in hash_groups.items()
    if len(paths) > 1
}


print("=" * 80)
print("CLEAN DATASET DUPLICATE CHECK")
print("=" * 80)

print(
    "Unique hashes:",
    len(hash_groups)
)

print(
    "Duplicate groups:",
    len(duplicates)
)
```

```python
leakage_groups = []


for h, paths in duplicates.items():

    splits_found = set()


    for path in paths:

        parts = path.parts


        if "train" in parts:

            splits_found.add("train")


        elif "valid" in parts:

            splits_found.add("valid")


        elif "test" in parts:

            splits_found.add("test")


    if len(splits_found) > 1:

        leakage_groups.append(
            (
                h,
                paths,
                splits_found
            )
        )


print("=" * 80)
print("FINAL CROSS-SPLIT LEAKAGE CHECK")
print("=" * 80)


print(
    "Cross-split duplicate groups:",
    len(leakage_groups)
)
```
